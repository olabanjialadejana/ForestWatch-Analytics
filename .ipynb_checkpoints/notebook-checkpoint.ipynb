```python
import geopandas as gpd
import ee
import geemap
```

```python
caroni = gpd.read_file('caroni.geojson')
```

```python
ee.Initialize()
```

```python
forest_watch_col = ee.Image("UMD/hansen/global_forest_change_2023_v1_11")
forest_watch_col
```

```python
print(forest_watch_col)
```

```python
forest_watch_col.visualize(min=0, max=255, palette=['red', '00FF00', 'blueviolet'])
```

```python
forest_watch_col.bandNames().getInfo()
```

```python
Map = geemap.Map()
```

```python
# Define a region of interest (e.g., a rectangle over Trinidad)
region = ee.Geometry.Rectangle([-62, 10, -60, 11])  # Adjust coordinates as needed

# Visualize forest loss (band: 'loss')
loss_vis_params = {'palette': ['white', 'red']}
Map.addLayer(forest_watch_col.select('loss'), loss_vis_params, 'Forest Loss')

# Visualize tree cover in 2000 (band: 'treecover2000')
treecover_vis_params = {'min': 0, 'max': 100, 'palette': ['white', 'green']}
Map.addLayer(forest_watch_col.select('treecover2000'), treecover_vis_params, 'Tree Cover 2000')

# Center the map on the region
Map.centerObject(region, zoom=8)
```
