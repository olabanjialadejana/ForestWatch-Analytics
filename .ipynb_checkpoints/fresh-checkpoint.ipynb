```python
import geemap
import geopandas as gpd
import json

def plot_geojson(geojson_path):
    """
    Plots the given GeoJSON file using geemap.

    Parameters:
    geojson_path (str): Path to the GeoJSON file.
    """
    # Load GeoJSON data
    with open('C:\Python Training\ForestWatch-Analytics\caroni.geojson', 'r') as f:
        geojson_data = json.load(f)

    # Convert to GeoDataFrame for bounds extraction
    gdf = gpd.GeoDataFrame.from_features(geojson_data['features'])

    # Create a map centered at the GeoJSON bounds
    bounds = gdf.total_bounds  # (minx, miny, maxx, maxy)
    center = [(bounds[1] + bounds[3]) / 2, (bounds[0] + bounds[2]) / 2]

    # Initialize geemap Map
    m = geemap.Map(center=center, zoom=10)

    # Add GeoJSON layer using geemap's built-in method
    m.add_geojson(geojson_path)

    # Display map
    return m



```

```python

# Plot and show the map

map_object = plot_geojson('C:\Python Training\ForestWatch-Analytics\caroni.geojson')
map_object.show()
```
