```python
import ee
import geemap

# Initialize Earth Engine
ee.Initialize()

# Load the Global Forest Change dataset
forest_watch = ee.Image("UMD/hansen/global_forest_change_2023_v1_11")

# Extract forest loss for 2014
loss_2014 = forest_watch.select('lossyear').eq(14)

# Visualize the data
vis_image = loss_2014.visualize(forceRgbOutput=True)

# Define a bounding box for Trinidad
trinidad_bbox = ee.Geometry.Rectangle([-61.95, 10.0, -60.45, 11.35])

# Create a map
m = geemap.Map()

# Add layers to the map
m.addLayer(vis_image, {}, "Forest Loss 2014")
m.addLayer(trinidad_bbox, {'color': 'red'}, "Trinidad Boundary")

# Center the map on Trinidad
m.centerObject(trinidad_bbox, zoom=8)

# Display the map
m
```
