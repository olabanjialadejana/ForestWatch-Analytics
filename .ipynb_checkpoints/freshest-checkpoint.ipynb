```python
import ee
import geemap
import json

# Authenticate and initialize Earth Engine
try:
    ee.Initialize()
except Exception as e:
    ee.Authenticate()
    ee.Initialize()

# Load GeoJSON AOI
geojson_path = "C:\Python Training\ForestWatch-Analytics\caroni2.geojson"  # 🔹 Update this path with your actual GeoJSON file
with open(geojson_path) as f:
    geojson = json.load(f)

# Convert GeoJSON to an Earth Engine Geometry
aoi = ee.Geometry(geojson['features'][0]['geometry'])

# Load Hansen Global Forest Change dataset and clip to AOI
hansen = ee.Image("UMD/hansen/global_forest_change_2023_v1_11").clip(aoi)

# Define a time range (2001-2023)
years = list(range(2001, 2024))

# Function to create an image for each year showing loss intensity
def calculate_loss(year):
    loss_mask = hansen.select('lossyear').eq(year - 2000)
    loss_area = loss_mask.multiply(ee.Image.pixelArea()).rename('loss_area')
    return loss_area.set('year', year)

# Generate a collection of annual forest loss images
loss_by_year = ee.ImageCollection([calculate_loss(year) for year in years])

# Sum the loss areas over time
total_loss = loss_by_year.sum().rename('total_loss')

# Define visualization parameters
vis_params = {
    'min': 0,
    'max': 50000,  # Adjust based on study region
    'palette': ['yellow', 'orange', 'red']
}

# Create a map
Map = geemap.Map()
Map.centerObject(aoi, 8)  # 🔹 Zoom into AOI
Map.addLayer(total_loss, vis_params, 'Annual Forest Loss Intensity (Clipped)')

# Display the map
Map
```
